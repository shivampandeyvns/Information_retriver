```python
from indexing import build_inverted_index
from indexing import search_inverted_index
from indexing import search_and_query
from indexing import search_or_query


documents = {
    "doc1": ["machine", "learning", "python"],
    "doc2": ["machine", "vision"],
    "doc3": ["python", "analytics"]
}

index = build_inverted_index(documents)

print("INDEX")
print(index)

print("\nSingle Term")
print(search_inverted_index("machine", index))

print("\nAND Query")
print(search_and_query("machine python", index))

print("\nOR Query")
print(search_or_query("machine python", index))
```

```python
from phrase_query import (
    build_biword_index,
    build_positional_index,
    search_biword,
    search_positional_phrase
)

documents = {

    "doc1": [
        "machine",
        "learning",
        "is",
        "fun"
    ],

    "doc2": [
        "machine",
        "vision",
        "and",
        "deep",
        "learning"
    ],

    "doc3": [
        "python",
        "machine",
        "learning"
    ]
}

# -------------------------
# BIWORD
# -------------------------

biword_index = build_biword_index(documents)

print("BIWORD INDEX")
print(biword_index)

print("\nBiword Search")
print(
    search_biword(
        "machine learning",
        biword_index
    )
)

# -------------------------
# POSITIONAL
# -------------------------

positional_index = build_positional_index(documents)

print("\nPositional Search")

print(
    search_positional_phrase(
        "machine learning",
        positional_index
    )
)
```
